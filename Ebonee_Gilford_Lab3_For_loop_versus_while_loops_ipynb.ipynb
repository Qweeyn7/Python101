# **Introduction**

### Both **for loops** and **while loops** are used to **execute a block of code multiple times**, but they are used in slightly different scenarios based on the nature of the condition that controls the loop




#  **1. For Loops!**

### Use a ***for loop when you know in advance how many times you need to execute the loop***. This is common in cases where you iterate over a sequence (like a list, tuple, or string) or when you want to run a loop a specific number of times. for loops are typically used for iterating over elements of a collection (e.g., each item in a list) or for running a loop a known number of times because the range is specified

### Example uses:
###- Iterating through elements in a list, tuple, or string.
###- Running a loop a fixed number of times using a range.

## **Example of using For loops**

### Run the following examples:

```python
# Iterating over a list
for item in [1, 2, 3, 4, 5]:
    print(item)


```

```python
# Running a loop a fixed number of times
for i in range(20): # 0 to 4
    print(i)
```

#  **2. While Loops!**

### Use a ***while loop when the number of times the loop needs to execute is not known in advance and depends on some condition being true***. It's more flexible in terms of controlling the loop since the condition can be based on dynamic data that changes as the loop executes. while loops are typically used for repeating a block of code as long as a certain condition remains true.

### Example uses:
### - Waiting for a specific condition to become False.
### - Running a loop until an external condition is met (e.g., user input, file content)

## **Example of using while loops**

### Run the following examples:

##  **Exercise: Guess the number!**

```python
# Looping until a condition is met
count = 0
while count < 5:
    print(count)
    count += 1  # Important to modify the condition variable within the loop
```
